```python
#Lendo tabela do volume com spark, poderia usar pandas,csv? sim, porém estou treinando usar spark.
df = spark.read.csv(
      "/Volumes/e_commerce_2026/bronze/volume/pedidos_carga_inicial.csv",
      header=True,
      inferSchema=True
)

display(df)
```

```python
#Codigo para ler tipo dos dados
df.printSchema()
```

```python
#Contando numero de linhas da tabela
df.count()
```

```python
#salvando na tabela bronze usando delta
#transformando a primeira tabela em delta e salvando no bronze. 1/2.
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("e_commerce_2026.bronze.e_commerce_bronze")
                                     
```

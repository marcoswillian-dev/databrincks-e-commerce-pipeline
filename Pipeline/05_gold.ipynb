```python
%sql
CREATE OR REPLACE TABLE e_commerce_2026.gold.vendas_por_categoria AS
SELECT
  category,
  status,
  COUNT(*) AS total_pedidos,
  SUM(total_price) AS receita_total
FROM e_commerce_2026.silver.pedidos_silver
GROUP BY category, status;
```

```python
%sql
SELECT * FROM e_commerce_2026.gold.vendas_por_categoria
ORDER BY receita_total DESC;
```
